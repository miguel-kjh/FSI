```python
import cv2
import os
```

```python
FOLDER_IMG = 'img/'
IMG = "road.jpg"
path_img = os.path.join(FOLDER_IMG, IMG)
print(path_img)
```

```python
img = cv2.imread(path_img)
print(type(img))
print(img.shape)  # alto, ancho, canales
```

```python
img
```

```python
img = cv2.imread(path_img)
cv2.imshow('image', img)
cv2.waitKey(0)
```

```python
img = cv2.imread(path_img)
(B, G, R) = img[100, 100]

print("R = {}, G = {}, B = {}".format(R, G, B))

B = img[100, 100, 0]
print("B = {}".format(B))
```

```python
img = cv2.imread(path_img)
print(img.shape)
roi = img[360:430, 441:500]
cv2.imshow("ROI", roi)
cv2.waitKey(0)
```

```python
dim = 500
resize = cv2.resize(img, (dim, dim))
cv2.imshow("Resize", resize)
cv2.waitKey(0)
```

```python
img = cv2.imread(path_img)
img_copy = img.copy()

rect = cv2.rectangle(
    img_copy, 
    (441, 360), # esquina superior izquierda
    (500, 430), # esquina inferior derecha
    (0, 255, 0), # color (B, G, R)
    2 # grosor del rectángulo
)
cv2.imshow("Rectangle", img_copy)
cv2.waitKey(0)
```

Guardar la imagen

```python
# Guardar la imagen
path_to_save = os.path.join(FOLDER_IMG, "road_car_detected.png")
cv2.imwrite(path_to_save, img_copy)
```

```python
img = cv2.imread(path_img)
img_copy = img.copy()

text = cv2.putText(
    img_copy, # imagen donde se escribe
    'OpenCV Demo', # texto
    (30, 100), # posición (x, y) de la esquina inferior izquierda del texto
    cv2.FONT_HERSHEY_SIMPLEX, # fuente
    4, # escala de la fuente
    (255, 0, 0), # color (B, G, R)
    2 # grosor del texto
)
cv2.imshow("Text", img_copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
